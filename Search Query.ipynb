```python
import pandas as pd
import re
from nltk.tokenize import word_tokenize
from nltk.metrics import edit_distance
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

#Load the dataset
df=pd.read_csv("spelling_corpus.csv")
print(df.head())
print(df.columns)
print("Number of words:",len(df))

#build vocabulary of correctly spelled words
vocabulary=set(df['correct_word'].dropna().astype(str).str.lower())
print("Vocabulary size:",len(vocabulary))
print(list(vocabulary)[:20])

#Accept user search query
query=input("enter your search query: ")
print("Original Query:",query)

#tokenize the query
tokens=word_tokenize(query.lower())
print("Tokens:",tokens)

#Identify incorrect words
incorrect_words = []

for word in tokens:
    if word not in vocabulary:
        incorrect_words.append(word)

print("Incorrect Words:", incorrect_words)
#calculate edit distance
def find_closest_word(word, vocabulary):
    distances = {}

    for candidate in vocabulary:
        distances[candidate] = edit_distance(word, candidate)

    closest_word = min(distances, key=distances.get)

    return closest_word, distances[closest_word]

#select the closest matching word
word = "machne"

suggestion, distance = find_closest_word(word, vocabulary)

print("Incorrect Word:", word)
print("Suggested Word:", suggestion)
print("Edit Distance:", distance)

#correct the complete query
corrected_tokens = []
for word in tokens:
    if word in vocabulary:
        corrected_tokens.append(word)
    else:
        suggestion, distance = find_closest_word(word, vocabulary)
        corrected_tokens.append(suggestion)
corrected_query = " ".join(corrected_tokens)
print("Corrected Query:", corrected_query)

#display required output
print("\n========== SEARCH QUERY SPELLING CORRECTOR ==========")

print("Original Query:")
print(query)

print("\nIncorrect Words:")

for word in incorrect_words:
    print(word)

print("\nSuggested Corrections:")

for word in incorrect_words:
    suggestion, distance = find_closest_word(word, vocabulary)

    print(word, "->", suggestion,
          "(Edit Distance:", distance, ")")

print("\nFinal Corrected Query:")
print(corrected_query)
```
